### A mock Computational Social science experiment
##### Research question: is people good at guessing the actual temperature in town?

Let' interview a random sample of walkers-by to collect their best gueess about what the temperature is like today.

Of course, we know the exact temperature at the time of interview, which ws assume constant over the whole experiment

Next, compute average and variance of the sample.

How's the sample average? Is the public over- or under-estimating temperature?

Essentially, how large is the variance? Can we operationalise this type of opinion aggregation?



```python
TEMPERATURE = 24

replies = [20, 25, 26, 24, 25, 21, 22, 27, 23, 23] 
```

```python
# count how many interviews we collected
n = 0

# internal variable, no meaning for the experiment
accumulator = 0

for estimate in replies:
    accumulator += estimate
    n += 1

# avoid division by 0!
if n > 0 :

    # now we can calculate the average of estimates
    sample_average = accumulator / n

    print('The average of the collected interviews is', sample_average)

    deviation = sample_average - TEMPERATURE

    print('Overall, the deviation from the thermometer is', deviation)

    if sample_average > TEMPERATURE:
        print('the public seems to over-estimate the temperature')
    else:
        print('the public seems to under-estimate the temperature in town')

    # fine, but what if the public is getting temp. just right?
    # Think how you could extend this code to treat the case of very low deviation between the two
   
else:
    print('Sorry but there were no respondents')
```
