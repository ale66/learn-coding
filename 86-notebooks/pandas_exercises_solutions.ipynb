````python
# Check that your kernel knows Pandas (and Numpy)

import numpy as np
import pandas as pd
````

##### Pandas Series

##### Create a series with values 35, 45, 72 at index positions A, B and C

````python
#you could use a dict

s = pd.Series({'A': 35, 'B':45, 'C':72})

s
````

````python
#pandas is built on top of matplotlib
#you call plotting functions directly

s.plot(color='green', title = 'series s')
````

##### Assign to the index the name `letters`

````python
# hint: set name of index

s.index.name = 'letters'

s
````

##### Assign to the Series the name `numbers`

````python
#set name of the series

s.name = 'numbers'

s
````

##### Turn the series into a Dataframe (DF) with 2 columns. 

The DataFrame should be indexed by numbers starting from 0.

The columns should be named 'Col1' and 'Col2'


````python
# hint: Pandas series are always horizontal, but we can transpose (.T)

df = pd.DataFrame([s.values,s.index]).T

df
````

````python
df.columns = ['Col1', 'Col2']

df
````

##### Retrieve the data in the second row

````python
df.iloc[1,:]
````

````python
# alternatively:
 
df.loc[1]
````

##### Retrieve data corresponding to column `Col2` 

````python
df.iloc[:,1]
````

````python
# alternatively

df['Col2']
````

##### Add a column, call it `added`, that takes value `100` for every row

````python
# hint: you do not have to repeat 100

df['added'] = 100

df
````

##### Retrieve values of the DF for the rows where the value of index numbers is even (0, 2, 4, ...)

````python
#you can not only index data, but use :: to set the pace of retrieval: from:to:step 

df.loc[0::2]
````

##### Retrieve data on the first row of column `Col1`

````python
df.iloc[0,0]
````

##### How many times does value 100 appears in the added column?

````python
df.added.value_counts()
````
